# Data Exploration using SQLite and Pandas Dataframe


**Importing Libraries**

```python
import pandas as pd
import numpy as np
import sqlite3
```

**Importing Dataset**

```python
data = pd.read_csv('Airbnb.csv')
data.head()
```

**Connecting sqlite database**

```python
conn = sqlite3.connect('airbnb_database.db')#this is database name
```

```python
#now lets make an table and import all data in sqlite3
# if exist mean if before table hai to replace and also index false mean index ko load mat kro wrna ek extra column add
data.to_sql('airbnb_table', conn, if_exists='replace', index=False)
```

```python
query = 'SELECT * From airbnb_table'
pd.read_sql(query, conn)
```

```python
query = 'SELECT * FROM airbnb_table'
pd.read_sql(query, conn)
```

**WHERE Room type = Private room**

```python
query = 'SELECT * FROM airbnb_table WHERE room_type = "Private room"'
pd.read_sql(query, conn)
```

**query to find neighbourhood = Mahanagar**

```python
query = 'SELECT * FROM airbnb_table WHERE neighbourhood = "Mahanagar" '
pd.read_sql(query, conn)
```
